```python
def f(x):
    return -(x-3)**2 + 10
```

```python
def Alt(x,dx):
    a=f(x-dx)
    b=f(x+dx)
    print(f"x={x}, f(x)={f(x)}, x-dx={x-dx}, f(x-dx)={a},x+dx={x+dx}, f(x+dx)={b}")
    if a > b:
        x = x - dx
    else:
        x = x + dx
    return x
```

```python
dx=1
x=1    
for i in range(10):
    x = Alt(x,dx)
```
