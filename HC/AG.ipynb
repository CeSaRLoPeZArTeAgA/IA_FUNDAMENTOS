```python
import numpy as np
def f(W,V,I):
    if((np.array(W)*np.array(I)).sum() > C_max):
        result= 0
    else:
        result= (np.array(W)*np.array(V)*np.array(I)).sum()
    print(f"peso={(np.array(W)*np.array(I)).sum()}, Fitness={result}    ")
    return result

#pesos
W=[2,3,4,5,9]
#valores
V=[3,4,8,8,10]
#valor maximo de peso
C_max=15

I=np.zeros((4,5))
p=np.zeros(4)

#Poblacion inicial
I[0]=[1, 0, 1, 0, 1]
I[1]=[0, 1, 1, 1, 0]
I[2]=[1, 1, 1, 1, 0]
I[3]=[0, 0, 0, 1, 1]

sum=0
for i in range(4):
    p[i]=f(W,V,I[i])
    sum=p[i]+sum

print(f"Fitness total={sum}")
p=p/sum
print(f"Probabilidades={p}")

#f(W,V,I[0])
#print(f(W,V,I[0]))
```
