```python
# 2. Dado el siguiente pseudocódigo que implementa un algoritmo de búsqueda, ¿qué tipo de búsqueda se está realizando?
def busqueda(arr, elemento):
  inicio = 0
  fin = arr.length() - 1
  while inicio <= fin:
    medio = (inicio + fin) / 2
    if arr[medio] == elemento:
      return medio
    if arr[medio] < elemento:
      inicio = medio + 1
    else:
      fin = medio - 1
    return -1

# b) Búsqueda binaria
```
