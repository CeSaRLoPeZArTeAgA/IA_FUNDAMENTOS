### Problemas de Linear Discriminant Analysis (LDA)

#### LDA-T1: Demostración de $S_W$ y $S_B$

**Objetivo:** Demostrar la descomposición $S_T = S_W + S_B$.

**Paso 1: Definición de la Desviación Total**
La matriz de dispersión total $S_T$ se define como:
$$S_T = \sum_{i=1}^N (\mathbf{x}_i - \boldsymbol{\mu}) (\mathbf{x}_i - \boldsymbol{\mu})^T$$

**Paso 2: Introducción de la Media de Clase**
Para cualquier punto $\mathbf{x}_i \in C_k$, podemos introducir la media de su clase $\boldsymbol{\mu}_k$:
$$(\mathbf{x}_i - \boldsymbol{\mu}) = (\mathbf{x}_i - \boldsymbol{\mu}_k) + (\boldsymbol{\mu}_k - \boldsymbol{\mu})$$

**Paso 3: Expansión del Producto Exterior**
Sustituimos esta identidad en la fórmula de $S_T$:
$$S_T = \sum_{k=1}^C \sum_{\mathbf{x}_i \in C_k} [(\mathbf{x}_i - \boldsymbol{\mu}_k) + (\boldsymbol{\mu}_k - \boldsymbol{\mu})] [(\mathbf{x}_i - \boldsymbol{\mu}_k) + (\boldsymbol{\mu}_k - \boldsymbol{\mu})]^T$$Expandimos el producto:$$S_T = \sum_{k=1}^C \sum_{\mathbf{x}_i \in C_k} \left[ (\mathbf{x}_i - \boldsymbol{\mu}_k)(\mathbf{x}_i - \boldsymbol{\mu}_k)^T + (\boldsymbol{\mu}_k - \boldsymbol{\mu})(\boldsymbol{\mu}_k - \boldsymbol{\mu})^T + 2(\mathbf{x}_i - \boldsymbol{\mu}_k)(\boldsymbol{\mu}_k - \boldsymbol{\mu})^T \right]$$

**Paso 4: Anulación del Término Cruzado**
El término cruzado es cero porque $\sum_{\mathbf{x}_i \in C_k} (\mathbf{x}_i - \boldsymbol{\mu}_k) = \mathbf{0}$:
$$\sum_{\mathbf{x}_i \in C_k} (\mathbf{x}_i - \boldsymbol{\mu}_k)(\boldsymbol{\mu}_k - \boldsymbol{\mu})^T = \left[ \sum_{\mathbf{x}_i \in C_k} (\mathbf{x}_i - \boldsymbol{\mu}_k) \right] (\boldsymbol{\mu}_k - \boldsymbol{\mu})^T = \mathbf{0}$$

**Paso 5: Conclusión de $S_W$ y $S_B$**
Los dos términos restantes definen $S_W$ (dispersión dentro de la clase) y $S_B$ (dispersión entre clases):

1.  **Matriz Intraclase ($S_W$):**
    $$S_W = \sum_{k=1}^C \sum_{\mathbf{x}_i \in C_k} (\mathbf{x}_i - \boldsymbol{\mu}_k) (\mathbf{x}_i - \boldsymbol{\mu}_k)^T \quad \text{(1)}$$
2.  **Matriz Entreclases ($S_B$):** Como $(\boldsymbol{\mu}_k - \boldsymbol{\mu})$ es constante para todos los $N_k$ puntos de $C_k$:
    $$S_B = \sum_{k=1}^C N_k (\boldsymbol{\mu}_k - \boldsymbol{\mu}) (\boldsymbol{\mu}_k - \boldsymbol{\mu})^T \quad \text{(2)}$$
    **Resultado:** $S_T = S_W + S_B$.

-----

#### LDA-T2: Demostración de la Razón de Rayleigh

**Objetivo:** Demostrar que LDA maximiza la razón $J(\mathbf{w}) = \frac{\mathbf{w}^T S_B \mathbf{w}}{\mathbf{w}^T S_W \mathbf{w}}$.

**Paso 1: Dispersión Proyectada**
La varianza (dispersión) de un conjunto de datos $\mathbf{x}$ con matriz de dispersión $S$ proyectado en una dirección $\mathbf{w}$ es $\mathbf{w}^T S \mathbf{w}$.

  * Dispersión entre clases proyectada: $\mathbf{w}^T S_B \mathbf{w}$.
  * Dispersión intraclase proyectada: $\mathbf{w}^T S_W \mathbf{w}$.

**Paso 2: Criterio de Fisher**
El criterio de Fisher busca maximizar la separación de las medias de clase **relativa** a la dispersión interna de las clases.
$$\max_{\mathbf{w}} \frac{\text{Separación Proyectada}}{\text{Dispersión Interna Proyectada}}$$

**Paso 3: Formulación Matemática**
Esto conduce directamente a la función $J(\mathbf{w})$, conocida como la Razón de Rayleigh o el Criterio de Fisher:
$$J(\mathbf{w}) = \frac{\mathbf{w}^T S_B \mathbf{w}}{\mathbf{w}^T S_W \mathbf{w}} \quad \text{(3)}$$

-----

#### LDA-T3: Demostración del Sistema de Autovalores Generalizado

**Objetivo:** Derivar $S_B \mathbf{w} = \lambda S_W \mathbf{w}$ a partir de la optimización de $J(\mathbf{w})$.

**Paso 1: Derivada y Condición de Máximo**
Para encontrar el máximo de $J(\mathbf{w})$ (Ec. 3), tomamos la derivada parcial con respecto a $\mathbf{w}$ y la igualamos a cero: $\nabla_{\mathbf{w}} J(\mathbf{w}) = \mathbf{0}$.
Usando la regla del cociente para derivadas matriciales:
$$\nabla_{\mathbf{w}} J(\mathbf{w}) = \frac{(\mathbf{w}^T S_W \mathbf{w}) \nabla_{\mathbf{w}} (\mathbf{w}^T S_B \mathbf{w}) - (\mathbf{w}^T S_B \mathbf{w}) \nabla_{\mathbf{w}} (\mathbf{w}^T S_W \mathbf{w})}{(\mathbf{w}^T S_W \mathbf{w})^2} = \mathbf{0}$$

**Paso 2: Simplificación del Numerador**
Para que la fracción sea cero, el numerador debe ser cero. Usamos $\nabla_{\mathbf{w}} (\mathbf{w}^T A \mathbf{w}) = 2A\mathbf{w}$ (para $A$ simétrica):
$$(\mathbf{w}^T S_W \mathbf{w}) (2 S_B \mathbf{w}) - (\mathbf{w}^T S_B \mathbf{w}) (2 S_W \mathbf{w}) = \mathbf{0}$$

**Paso 3: Rearreglo y Sustitución**
Dividimos por 2 y reordenamos:
$$(\mathbf{w}^T S_W \mathbf{w}) S_B \mathbf{w} = (\mathbf{w}^T S_B \mathbf{w}) S_W \mathbf{w}$$Dividimos por el escalar $(\mathbf{w}^T S_W \mathbf{w})$ (asumiendo que $S_W$ es no singular y, por lo tanto, el denominador no es cero):$$S_B \mathbf{w} = \frac{\mathbf{w}^T S_B \mathbf{w}}{\mathbf{w}^T S_W \mathbf{w}} S_W \mathbf{w}$$

**Paso 4: Identificación del Autovalor**
Sustituimos la razón de Fisher $J(\mathbf{w}) = \lambda$:
$$S_B \mathbf{w} = \lambda S_W \mathbf{w} \quad \text{(4)}$$
Este es el sistema de autovalores generalizado.

### Problemas de t-distributed Stochastic Neighbor Embedding (t-SNE)

#### t-SNE-T4: Similitudes en Alta Dimensión y Simetrización

**Objetivo:** Demostrar la simetrización de la probabilidad condicional.

**Paso 1: Probabilidad Condicional ($p_{j|i}$)**
Utiliza un kernel Gaussiano centrado en $\mathbf{x}_i$, con una varianza $\sigma_i$ que se ajusta por *Perplexity*:
$$p_{j|i} = \frac{\exp(-\| \mathbf{x}_i - \mathbf{x}_j \|^2 / 2\sigma_i^2)}{\sum_{k \neq i} \exp(-\| \mathbf{x}_i - \mathbf{x}_k \|^2 / 2\sigma_i^2)} \quad \text{(5)}$$
$p_{j|i}$ es la probabilidad de que $\mathbf{x}_i$ elija a $\mathbf{x}_j$ como su vecino.

**Paso 2: La Necesidad de Simetría**
Como $\sigma_i$ varía con $i$, $p_{j|i} \neq p_{i|j}$, lo que crea una distribución asimétrica. Para minimizar la Divergencia de Kullback-Leibler $D_{KL}(P||Q)$, necesitamos una distribución $P$ conjunta y simétrica.

**Paso 3: Simetrización**
La probabilidad conjunta $p_{ij}$ se define como el promedio simétrico:
$$p_{ij} = \frac{p_{j|i} + p_{i|j}}{2N} \quad \text{o, en la práctica:} \quad p_{ij} = \frac{p_{j|i} + p_{i|j}}{2 \sum_{k \neq l} p_{l|k}} \quad \text{(6)}$$
Donde la normalización asegura $\sum_{i,j} p_{ij} = 1$.

-----

#### t-SNE-T5: Kernel de Cauchy y *Crowding Problem*

**Objetivo:** Explicar el uso de la distribución t de Student.

**Paso 1: Probabilidad de Similitud en Baja Dimensión ($q_{ij}$)**
En baja dimensión (espacio $Y$), t-SNE usa una distribución **t de Student con un grado de libertad** (kernel de Cauchy), que tiene la forma:
$$q_{ij} = \frac{(1 + \| \mathbf{y}_i - \mathbf{y}_j \|^2)^{-1}}{\sum_{k \neq l} (1 + \| \mathbf{y}_k - \mathbf{y}_l \|^2)^{-1}} \quad \text{(7)}$$

**Paso 2: Mitigación del *Crowding Problem***

  * **El Problema:** Al mapear un vecindario de $D$ dimensiones a 2D, se necesita un espacio proyectado muy grande si se usa un kernel Gaussiano en $Q$. Los puntos ligeramente distantes en alta D terminan apiñados o "abarrotados" (crowding) alrededor del centro en baja D.
  * **La Solución (Colas Pesadas):** La distribución t de Student tiene **colas mucho más pesadas** que una Gaussiana.
      * **Implicación:** Esto significa que las distancias moderadas en el espacio $Y$ (baja dimensión) son penalizadas menos que por una Gaussiana. Esto proporciona una fuerza de repulsión más fuerte a distancias grandes, lo que permite que los clústeres se separen en 2D de manera más efectiva, aliviando el *crowding problem*.

-----

#### t-SNE-T6: Derivación del Gradiente

**Objetivo:** Derivar el gradiente de la función de costo $C = D_{KL}(P || Q)$.

**Paso 1: Gradiente de KL-Divergence**
La función de costo es $C = \sum_{i} \sum_{j \neq i} p_{ij} \log \left( \frac{p_{ij}}{q_{ij}} \right)$.
La derivada con respecto a $\mathbf{y}_i$ (utilizando la regla de la cadena y que $\sum_{j} q_{ij}$ y $\sum_{k,l} q_{kl}$ son funciones implícitas de $\mathbf{y}_i$) resulta en:

**Paso 2: La Fórmula del Gradiente**
$$\frac{\partial C}{\partial \mathbf{y}_i} = 4 \sum_{j \neq i} (p_{ij} - q_{ij}) (1 + \| \mathbf{y}_i - \mathbf{y}_j \|^2)^{-1} (\mathbf{y}_i - \mathbf{y}_j) \quad \text{(8)}$$

**Paso 3: Interpretación (Fuerzas)**

  * **Atracción:** Si $p_{ij} > q_{ij}$ (puntos cercanos en alta D están lejos en baja D), el término $(p_{ij} - q_{ij})$ es positivo, y el gradiente tiene la misma dirección que $(\mathbf{y}_i - \mathbf{y}_j)$. El algoritmo (Descenso de Gradiente) moverá $\mathbf{y}_i$ en la dirección opuesta, **atrayéndolo** hacia $\mathbf{y}_j$.
  * **Repulsión:** Si $p_{ij} < q_{ij}$ (puntos lejanos en alta D están cerca en baja D), el término es negativo. El algoritmo moverá $\mathbf{y}_i$ en la misma dirección que $(\mathbf{y}_i - \mathbf{y}_j)$, **repeliéndolo** de $\mathbf{y}_j$.

#### LDA-P3: Interpretación de Autovalores

**a) Dimensionalidad Óptima de Reducción**

**b) Importancia Relativa de Cada Eje Discriminante**

  * **Ejes 3, 4, 5 ($\lambda \approx 0$):** Estos autovalores insignificantes (o cero) indican que los ejes correspondientes no tienen capacidad discriminante. Son ruido o corresponden a direcciones donde las medias de clase ya no se separan.

### Problemas de t-distributed Stochastic Neighbor Embedding (t-SNE)

#### t-SNE-P4: Probabilidades Condicionales

**Contexto:** $\mathbf{x}_1=[0, 0]^T$, $\mathbf{x}_2=[1, 0]^T$, $\mathbf{x}_3=[10, 10]^T$. $\sigma_i=1$. $2\sigma_i^2=2$.

**Paso 1: Distancias Cuadradas (desde $\mathbf{x}_1$)**

  * $d_{12}^2 = \| \mathbf{x}_1 - \mathbf{x}_2 \|^2 = 1^2 = 1$
  * $d_{13}^2 = \| \mathbf{x}_1 - \mathbf{x}_3 \|^2 = 10^2 + 10^2 = 200$

**Paso 2: Términos Exponenciales (desde $\mathbf{x}_1$)**

  * $Num_{2|1} = \exp(-1/2) \approx 0.6065$
  * $Num_{3|1} = \exp(-200/2) = \exp(-100) \approx 3.72 \times 10^{-44}$

**Paso 3: Cálculo de Probabilidades ($p_{j|1}$, ref. Ec. 5)**
$$Denom_1 = Num_{2|1} + Num_{3|1} \approx 0.6065$$$$p_{2|1} = \frac{0.6065}{0.6065} \approx \mathbf{1.000}$$$$p_{3|1} = \frac{3.72 \times 10^{-44}}{0.6065} \approx \mathbf{0.000}$$

**Discusión:** La función kernel Gaussiana, especialmente para una $\sigma$ pequeña (como $\sigma=1$), tiene un decaimiento muy rápido. Incluso una distancia moderada (10 unidades) resulta en una probabilidad de vecindad casi nula. Esto confirma que t-SNE se enfoca en la **preservación de la estructura local e inmediata**.

-----

#### t-SNE-P5: Efecto de *Perplexity*

**El parámetro *Perplexity* controla la escala de vecindad que t-SNE intenta preservar.**

| Parámetro | Resultado Esperado | Explicación |
| :--- | :--- | :--- |
| **`perplexity = 5` (Bajo)** | Clústeres fragmentados y ruidosos. | Se enfoca en un vecindario muy pequeño, ignorando la estructura más amplia. Los puntos periféricos a veces forman sus propios "micro-clústeres". La estructura global es pobre. |
| **`perplexity = 80` (Alto)** | Clústeres fusionados o difusos. | El $\sigma_i$ se vuelve grande, haciendo que t-SNE considere a muchos puntos distantes como vecinos. Esto resulta en *oversmoothing* (suavizado excesivo) y la potencial fusión de clústeres que deberían estar separados. |
| **Recomendación** | **Balance** | Generalmente, se recomienda un valor entre 5 y 50 para la mayoría de los datasets. |

-----

#### t-SNE-P6: Fuerzas de Gradiente

**Contexto:** $p_{ij} = 0.8$ (Alto), $q_{ij} = 0.01$ (Bajo).
**Fórmula de la fuerza:** $\mathbf{F}_{ij} \propto (p_{ij} - q_{ij}) \cdot \text{término de repulsión/atracción} \cdot (\mathbf{y}_i - \mathbf{y}_j)$ (ref. Ec. 8)

**Análisis:**

1.  **Error (KL-Divergence):** $p_{ij} > q_{ij}$. Esto significa que **dos puntos que estaban cerca en alta D están lejos en baja D**. Esto contribuye a un valor alto en la función de costo $C$.
2.  **Factor de Magnitud:** $p_{ij} - q_{ij} = 0.8 - 0.01 = 0.79$ (Positivo).
3.  **Dirección de Optimización:** El proceso de Descenso de Gradiente actualiza $\mathbf{y}_i$ en la dirección $-\nabla C$.
      * Como $p_{ij} - q_{ij}$ es positivo, la fuerza aplicada por el gradiente apunta en la dirección de $\mathbf{y}_i - \mathbf{y}_j$ (es decir, alejando a $\mathbf{y}_i$ de $\mathbf{y}_j$).
      * Sin embargo, el Descenso de Gradiente (que es $\propto - \nabla C$) aplicará una fuerza en la dirección opuesta, es decir, **atrayendo a $\mathbf{y}_i$ hacia $\mathbf{y}_j$**.

**Conclusión:** Se aplica una fuerte **fuerza de ATRACCIÓN**. El algoritmo está corrigiendo el gran error donde la vecindad local se perdió en el mapeo de baja dimensión.

## Solucionario Etapa 3: Programación en Python para Google Colab

### LDA-C1: LDA y Clasificación con Dataset Wine

```python
# CELDAS DE CÓDIGO

# Asegúrate de ejecutar esta celda si estás en un entorno diferente a Colab
# !pip install scikit-learn matplotlib

import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Cargar y Preparar Datos (Wine Dataset)
print("Cargando y preparando el dataset Wine (13D, 3 clases)...")
wine = load_wine()
X = wine.data
y = wine.target
target_names = wine.target_names

# Dividir para Evaluación
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Escalar los datos (importante para LDA)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 2. Aplicar LDA (Reducción a d=2)
# d=2 es el máximo para C=3 clases (C-1).
lda = LinearDiscriminantAnalysis(n_components=2)
# LDA debe ajustarse (fit) con X_train Y y_train (es supervisado)
X_train_lda = lda.fit_transform(X_train_scaled, y_train)
X_test_lda = lda.transform(X_test_scaled)

print(f"\nDimensionalidad Original: {X.shape[1]}")
print(f"Dimensionalidad Reducida (LDA): {X_train_lda.shape[1]}")

# 3. Visualización de la Proyección 2D
# La visualización debe mostrar una separación clara y lineal de los 3 clústeres.
plt.figure(figsize=(10, 8))
colors = ['navy', 'turquoise', 'darkorange']
lw = 2

for color, i, target_name in zip(colors, [0, 1, 2], target_names):
    plt.scatter(X_train_lda[y_train == i, 0], X_train_lda[y_train == i, 1], alpha=0.8, color=color,
                label=target_name, lw=lw)

plt.legend(loc='best', shadow=False, scatterpoints=1)
plt.title('Proyección LDA del Dataset Wine (2 Componentes)')

# La varianza explicada debe ser muy alta, típicamente > 99.9% en total para estos dos ejes.
print(f"Varianza Explicada Total por 2 Componentes LDA: {lda.explained_variance_ratio_.sum():.4f}")
plt.xlabel(f'Componente Discriminante 1 (Varianza Explicada: {lda.explained_variance_ratio_[0]:.2f})')
plt.ylabel(f'Componente Discriminante 2 (Varianza Explicada: {lda.explained_variance_ratio_[1]:.2f})')
plt.grid(True)
plt.show()

# 4. Evaluación de Clasificación en Espacio Reducido
# La Regresión Logística debe lograr una precisión (Accuracy) casi perfecta debido a la separación lineal.
lr = LogisticRegression(solver='liblinear', random_state=42)
lr.fit(X_train_lda, y_train)
y_pred_lda = lr.predict(X_test_lda)
accuracy_lda = accuracy_score(y_test, y_pred_lda)

print("\n--- Evaluación de Clasificación (Espacio Reducido LDA) ---")
print(f"Precisión (Accuracy) con Regresión Logística en 2D: {accuracy_lda:.4f}")
```

### t-SNE-C2: t-SNE para Visualización con Dataset MNIST

```python
# CELDAS DE CÓDIGO

# !pip install scikit-learn matplotlib
import time
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import fetch_openml
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler
import pandas as pd # Para la carga de datos de fetch_openml

# 1. Cargar y Preparar Datos (MNIST Dataset)
print("Cargando el dataset MNIST...")
# Se usa parser='auto' para manejar la estructura de datos moderna de scikit-learn
mnist = fetch_openml('mnist_784', version=1, cache=True, parser='auto')
X = mnist.data.values.astype(float) # Se convierte a numpy array
y = mnist.target.values.astype(int)

# Tomar un subconjunto (5000 muestras)
N_SAMPLES = 5000
np.random.seed(42)
# Seleccionar índices aleatorios
idx = np.random.choice(len(X), N_SAMPLES, replace=False)
X_subset = X[idx]
y_subset = y[idx]

# Escalar los datos (importante para t-SNE)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_subset)

print(f"Dimensionalidad Original: {X_scaled.shape[1]}")
print(f"Número de Muestras: {N_SAMPLES}")

def run_tsne_and_plot(X_data, y_labels, perplexity_val, title_suffix):
    """Ejecuta t-SNE y grafica los resultados."""
    print(f"\nEjecutando t-SNE con Perplexity = {perplexity_val}...")
    time_start = time.time()

    # Aplicar t-SNE (Usar n_jobs=-1 para paralelización)
    tsne = TSNE(n_components=2,
                perplexity=perplexity_val,
                n_iter=1000,
                random_state=42,
                verbose=1,
                n_jobs=-1)

    X_tsne = tsne.fit_transform(X_data)

    print(f"t-SNE finalizado en {time.time() - time_start:.2f} segundos.")

    # Graficar
    # Se utiliza un mapa de colores 'Spectral' con 10 colores para los 10 dígitos (0-9).
    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y_labels, cmap=plt.cm.get_cmap('Spectral', 10), alpha=0.7)
    plt.colorbar(scatter, ticks=range(10), label='Dígitos')
    plt.clim(-0.5, 9.5)
    plt.title(f'Visualización t-SNE del Dataset MNIST (Perplexity={perplexity_val}) {title_suffix}')
    plt.xlabel('Dimensión 1')
    plt.ylabel('Dimensión 2')
    plt.grid(True, alpha=0.2)
    plt.show()

# 2. Experimentar con Perplexity = 10 (Estructura Local Fuerte)
run_tsne_and_plot(X_scaled, y_subset, 10, "(Enfoque Local Fuerte)")

# 3. Experimentar con Perplexity = 50 (Balance Local/Global)
run_tsne_and_plot(X_scaled, y_subset, 50, "(Estructura Balanceada)")

# Comentario de Salida Esperada:
# El gráfico con Perplexity=50 (estructura balanceada) debería mostrar clústeres más definidos y mejor separados,
# siendo generalmente la mejor representación visual para MNIST. El gráfico con Perplexity=10 (foco local)
# podría presentar clústeres más dispersos o fragmentados debido a la baja escala de vecindad.
```
