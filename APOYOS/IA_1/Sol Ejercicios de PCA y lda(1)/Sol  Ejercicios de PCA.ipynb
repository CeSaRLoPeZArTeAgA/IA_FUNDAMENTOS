### Soluciones Etapa 1 – Comprensión Teórica
----
#### T1: Derivación (Máxima Varianza $\rightarrow$ Autovalores)

**Objetivo:** Demostrar que el problema de maximizar la varianza proyectada bajo restricción de norma lleva al problema de autovalores.

**Paso 1: Definición de la Función Objetivo**
La varianza de los datos centrados proyectados sobre un vector unitario $\mathbf{u}$ es la forma cuadrática dada por la Ecuación (3):
$$\text{Var}(\mathbf{u}^T \tilde{X}) = \mathbf{u}^T C \mathbf{u} \quad (3)$$

**Paso 2: Formulación del Lagrangiano**
El problema es maximizar (3) sujeto a la restricción $\mathbf{u}^T \mathbf{u} = 1$. Utilizamos el multiplicador de Lagrange $\lambda$:
$$\mathcal{L}(\mathbf{u}, \lambda) = \mathbf{u}^T C \mathbf{u} - \lambda (\mathbf{u}^T \mathbf{u} - 1) \quad (4)$$

**Paso 3: Derivación e Igualación a Cero**
Para encontrar los extremos, tomamos la derivada parcial del Lagrangiano (4) con respecto al vector $\mathbf{u}$ y la igualamos al vector nulo $\mathbf{0}$:
$$\nabla_{\mathbf{u}} \mathcal{L} = \frac{\partial}{\partial \mathbf{u}} \left[ \mathbf{u}^T C \mathbf{u} - \lambda (\mathbf{u}^T \mathbf{u} - 1) \right] = \mathbf{0}$$Aplicando las reglas de derivación matricial:$$2 C \mathbf{u} - 2 \lambda \mathbf{u} = \mathbf{0}$$

**Paso 4: Conclusión**
Dividiendo por 2 y reordenando, obtenemos la ecuación canónica del problema de autovalores (2):
$$C \mathbf{u} = \lambda \mathbf{u} \quad (2)$$
Los vectores $\mathbf{u}$ que maximizan (o minimizan) la varianza proyectada son los autovectores de $C$, y los valores de la varianza proyectada son sus autovalores asociados $\lambda$. **Q.E.D.**

----
#### T2: Equivalencia PCA $\Leftrightarrow$ Descomposición

**Objetivo:** Demostrar que autovectores correspondientes a autovalores distintos de $C$ son ortogonales.

**Paso 1: Definición del Problema de Autovalores**
Consideramos dos autovectores $\mathbf{u}_1$ y $\mathbf{u}_2$ con autovalores distintos $\lambda_1 \neq \lambda_2$:

1.  $C \mathbf{u}_1 = \lambda_1 \mathbf{u}_1$
2.  $C \mathbf{u}_2 = \lambda_2 \mathbf{u}_2$

**Paso 2: Producto Interno del Primer Caso**
Multiplicamos (1) por $\mathbf{u}_2^T$:
$$\mathbf{u}_2^T C \mathbf{u}_1 = \mathbf{u}_2^T (\lambda_1 \mathbf{u}_1) = \lambda_1 (\mathbf{u}_2^T \mathbf{u}_1) \quad (*)$$

**Paso 3: Producto Interno del Segundo Caso**
Multiplicamos (2) por $\mathbf{u}_1^T$:
$$\mathbf{u}_1^T C \mathbf{u}_2 = \mathbf{u}_1^T (\lambda_2 \mathbf{u}_2) = \lambda_2 (\mathbf{u}_1^T \mathbf{u}_2) \quad (**)$$

**Paso 4: Uso de la Simetría de $C$**
Como $C$ es una matriz de covarianza, es simétrica ($C = C^T$). Usando la propiedad $\mathbf{a}^T B \mathbf{c} = (\mathbf{a}^T B \mathbf{c})^T = \mathbf{c}^T B^T \mathbf{a}$, se cumple que:
$$\mathbf{u}_2^T C \mathbf{u}_1 = (\mathbf{u}_2^T C \mathbf{u}_1)^T = \mathbf{u}_1^T C^T \mathbf{u}_2 = \mathbf{u}_1^T C \mathbf{u}_2$$

**Paso 5: Igualación y Conclusión**
Igualando $(*)$ y $(**)$ usando la simetría:
$$\lambda_1 (\mathbf{u}_2^T \mathbf{u}_1) = \lambda_2 (\mathbf{u}_1^T \mathbf{u}_2)$$Como $\mathbf{u}_2^T \mathbf{u}_1 = \mathbf{u}_1^T \mathbf{u}_2$, reordenamos:$$(\lambda_1 - \lambda_2) (\mathbf{u}_1^T \mathbf{u}_2) = 0$$Dado que asumimos $\lambda_1 \neq \lambda_2$, debe ser $\lambda_1 - \lambda_2 \neq 0$. Por lo tanto, el producto interno debe ser:$$\mathbf{u}_1^T \mathbf{u}_2 = 0$$
Lo que implica que los autovectores $\mathbf{u}_1$ y $\mathbf{u}_2$ son **ortogonales** (independientes), cumpliendo el requisito de PCA. **Q.E.D.**

----
#### T3: Relación PCA $\leftrightarrow$ SVD

**Objetivo:** Deducir la relación entre los CP (autovectores de $C$) y la SVD de $\tilde{X}$.

**Paso 1: Definición de $C$ mediante $\tilde{X}$**
Partimos de la definición de la Matriz de Covarianza (ignorando el factor $M-1$ ya que no afecta los autovectores):
$$C \propto \tilde{X}^T \tilde{X}$$

**Paso 2: Sustitución de la SVD de $\tilde{X}$**
Sustituimos la SVD (5) en la expresión de $\tilde{X}^T \tilde{X}$:
$$\tilde{X} = U \Sigma V^T \quad (5)$$$$C \propto (U \Sigma V^T)^T (U \Sigma V^T)$$$$C \propto (V \Sigma^T U^T) (U \Sigma V^T)$$

**Paso 3: Uso de la Propiedad Ortogonal de $U$**
$U$ es una matriz ortogonal, por lo que $U^T U = I$ (la matriz identidad):
$$C \propto V \Sigma^T (\underbrace{U^T U}_{I}) \Sigma V^T$$
$$C \propto V (\Sigma^T \Sigma) V^T$$

**Paso 4: Diagonal de $\Sigma^2$**
$\Sigma$ es diagonal (o pseudodiagonal) con los valores singulares $\sigma_i$. Por lo tanto, $\Sigma^T \Sigma = \Sigma^2$ es una matriz diagonal con $\sigma_i^2$ en la diagonal.
$$C \propto V \Sigma^2 V^T$$

**Paso 5: Conexión con Autovalores y Conclusión**
La expresión $C V = V \Sigma^2$ (multiplicando por $V$ a la derecha y usando $V^T V = I$) es la forma matricial de $C \mathbf{u}_i = \lambda_i \mathbf{u}_i$. Por lo tanto:

  * Las columnas de $V$ son los autovectores de $C$, es decir, los **Componentes Principales**.
  * Los autovalores $\lambda_i$ son proporcionales a $\sigma_i^2$. Considerando el factor $(M-1)$ de (1), se obtiene la relación (6):
    $$\lambda_i = \frac{\sigma_i^2}{M-1} \quad (6)$$
    **Conclusión:** PCA puede implementarse encontrando la SVD de $\tilde{X}$, lo cual es equivalente a la descomposición de autovalores de $C$.

#### T4: Minimización del Error de Reconstrucción

**Objetivo:** Demostrar que PCA minimiza el error de reconstrucción (Proyección Ortogonal).

**Paso 1: Definición de Proyección y Error**
La proyección $\tilde{X}_k$ en el subespacio de $k$ dimensiones definido por los $k$ componentes principales con mayor varianza (columnas de $V_k$) es $\tilde{X}_k = \tilde{X} V_k V_k^T$.
El error de reconstrucción (7) es:
$$\text{Error}_{\text{reconstrucción}} = \| \tilde{X} - \tilde{X}_k \|_F^2 = \| \tilde{X} - \tilde{X} V_k V_k^T \|_F^2 \quad (7)$$

**Paso 2: Utilizando la Descomposición de $C$**
La matriz $C$ es diagonalizable y puede expresarse en términos de sus $N$ autovectores $\mathbf{u}_i$ y autovalores $\lambda_i$:
$$C = \sum_{i=1}^N \lambda_i \mathbf{u}_i \mathbf{u}_i^T$$
El error de reconstrucción está inversamente relacionado con la varianza capturada. El error al proyectar sobre un subespacio de $k$ CP está dado por la varianza no capturada por ese subespacio.

**Paso 3: Conexión con Varianza No Explicada**
El error cuadrático de reconstrucción (7) se simplifica a la suma de los autovalores que **no** fueron seleccionados (Teorema de Eckart–Young):
$$\| \tilde{X} - \tilde{X}_k \|_F^2 = (M-1) \sum_{i=k+1}^N \lambda_i$$
Para minimizar el error, debemos elegir los $k$ componentes que dejen la menor suma de autovalores restantes $\sum_{i=k+1}^N \lambda_i$. Dado que los autovalores $\lambda_i$ son no negativos y están ordenados de forma descendente ($\lambda_1 \ge \lambda_2 \ge \dots$), la forma de minimizar la suma de los $N-k$ autovalores restantes es **seleccionando los $k$ autovalores más grandes**.

**Paso 4: Conclusión**
Dado que los $k$ componentes principales de PCA corresponden a los $k$ autovalores más grandes, PCA es la transformación lineal que minimiza el error de reconstrucción cuadrático medio, en el sentido de la aproximación de bajo rango de la matriz de datos. **Q.E.D.**

#### T5: Propiedades: Escalado y Rotación

**Objetivo:** Mostrar la invarianza ante rotación y la necesidad de estandarización.

**a) Invarianza de la Traza ante Rotación**

**Paso 1: Definición de la Nueva Covarianza $C'$**
Sea $\tilde{X}' = \tilde{X}R$ la matriz de datos rotada, donde $R$ es una matriz de rotación ortogonal ($R^T R = I$). La nueva matriz de covarianza $C'$ (usando $\propto$ para ignorar el factor $M-1$):
$$C' \propto (\tilde{X}' )^T \tilde{X}' = (\tilde{X}R)^T (\tilde{X}R) = R^T \tilde{X}^T \tilde{X} R$$
$$C' \propto R^T C R$$

**Paso 2: Aplicación de la Propiedad de la Traza**
La traza de una matriz es la suma de sus elementos diagonales. Una propiedad clave de la traza es que $\text{Trace}(AB) = \text{Trace}(BA)$.
$$\text{Trace}(C') \propto \text{Trace}(R^T C R)$$Aplicando la propiedad con $A = R^T$ y $B = CR$:$$\text{Trace}(C') \propto \text{Trace}(C R R^T)$$Dado que $R$ es ortogonal, $R R^T = I$:$$\text{Trace}(C') \propto \text{Trace}(C I) = \text{Trace}(C)$$
**Conclusión:** La varianza total de los datos ($\text{Trace}(C)$), que es la suma de los autovalores, permanece **invariante** ante una rotación ortogonal. Esto significa que la varianza total explicada por PCA no cambia, solo su distribución entre los ejes (los autovectores se rotan).

**b) Efecto del Escalado y Necesidad de Estandarización**

**Paso 1: Función Objetivo con Escalado Desigual**
La función objetivo de PCA es maximizar la varianza proyectada $\mathbf{u}^T C \mathbf{u}$ (3). Los elementos de $C$ reflejan las escalas originales de las variables.

**Paso 2: Impacto del Escalado Desigual**
Si la Característica 1 tiene una varianza de $1000$ y la Característica 2 una varianza de $1$ (escalas desiguales), la matriz $C$ estará dominada por el valor $1000$. El primer componente principal $\mathbf{u}_1$ estará forzado a alinearse casi exclusivamente con la dirección de la Característica 1, **independientemente de si esta característica tiene el contenido informativo más relevante**.

**Paso 3: Conclusión**
El escalado desigual hace que PCA se alinee con la dimensión de mayor **escala** (o rango de valores) en lugar de la dimensión de mayor **información**. La **estandarización** (restar la media y dividir por la desviación estándar) asegura que todas las características contribuyan con una varianza inicial de 1, evitando que características con unidades grandes dominen el análisis.

### Soluciones Etapa 2 – Prácticos en Papel

#### P1: Cálculo Básico de $C$ y Proyección

**Datos:** $X = \begin{pmatrix} 2 & 1 \\ 4 & 3 \\ 6 & 2 \end{pmatrix}$, $M=3$.

**Paso 1: Calcular el Vector de Medias $\mathbf{\bar{x}}$**
$$\mathbf{\bar{x}} = \begin{pmatrix} \frac{2+4+6}{3} \\ \frac{1+3+2}{3} \end{pmatrix} = \begin{pmatrix} \frac{12}{3} \\ \frac{6}{3} \end{pmatrix} = \begin{pmatrix} 4 \\ 2 \end{pmatrix}$$

**Paso 2: Calcular la Matriz de Datos Centrados $\tilde{X}$**
$$\tilde{X} = X - \mathbf{1}\mathbf{\bar{x}}^T = \begin{pmatrix} 2-4 & 1-2 \\ 4-4 & 3-2 \\ 6-4 & 2-2 \end{pmatrix} = \begin{pmatrix} -2 & -1 \\ 0 & 1 \\ 2 & 0 \end{pmatrix}$$

**Paso 3: Calcular la Matriz de Covarianza $C$**
Utilizamos la Ecuación (1) con $M-1 = 2$:
$$C = \frac{1}{M-1} \tilde{X}^T \tilde{X} = \frac{1}{2} \begin{pmatrix} -2 & 0 & 2 \\ -1 & 1 & 0 \end{pmatrix} \begin{pmatrix} -2 & -1 \\ 0 & 1 \\ 2 & 0 \end{pmatrix}$$
$$C = \frac{1}{2} \begin{pmatrix} (4+0+4) & (2+0+0) \\ (2+0+0) & (1+1+0) \end{pmatrix} = \frac{1}{2} \begin{pmatrix} 8 & 2 \\ 2 & 2 \end{pmatrix} = \begin{pmatrix} 4 & 1 \\ 1 & 1 \end{pmatrix}$$

**Resultado:**
$$\mathbf{\bar{x}} = \begin{pmatrix} 4 \\ 2 \end{pmatrix}, \quad \tilde{X} = \begin{pmatrix} -2 & -1 \\ 0 & 1 \\ 2 & 0 \end{pmatrix}, \quad C = \begin{pmatrix} 4 & 1 \\ 1 & 1 \end{pmatrix}$$

#### P2: Autovalores y Autovectores (2x2)

**Datos:** $C = \begin{pmatrix} 5 & 1 \\ 1 & 5 \end{pmatrix}$.

**Paso 1: Resolver el Polinomio Característico**
Para hallar los autovalores $\lambda$, resolvemos $\det(C - \lambda I) = 0$:
$$\det \begin{pmatrix} 5-\lambda & 1 \\ 1 & 5-\lambda \end{pmatrix} = 0$$$$(5-\lambda)(5-\lambda) - 1(1) = 0$$$$\lambda^2 - 10\lambda + 25 - 1 = 0$$
$$\lambda^2 - 10\lambda + 24 = 0$$
Factorizando: $(\lambda - 6)(\lambda - 4) = 0$.
**Autovalores:** $\lambda_1 = 6$ y $\lambda_2 = 4$.

**Paso 2: Hallar el Primer Autovector $\mathbf{u}_1$ (asociado a $\lambda_1 = 6$)**
Usamos la Ecuación (2): $C \mathbf{u}_1 = \lambda_1 \mathbf{u}_1$, o $(C - \lambda_1 I) \mathbf{u}_1 = \mathbf{0}$.
$$\begin{pmatrix} 5-6 & 1 \\ 1 & 5-6 \end{pmatrix} \begin{pmatrix} u_{1,1} \\ u_{1,2} \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix} \quad \Rightarrow \quad \begin{pmatrix} -1 & 1 \\ 1 & -1 \end{pmatrix} \begin{pmatrix} u_{1,1} \\ u_{1,2} \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$
La primera fila da: $-u_{1,1} + u_{1,2} = 0 \quad \Rightarrow \quad u_{1,1} = u_{1,2}$.
Un autovector no normalizado es $\begin{pmatrix} 1 \\ 1 \end{pmatrix}$. Normalizando:
$$\mathbf{u}_1 = \frac{1}{\sqrt{1^2+1^2}} \begin{pmatrix} 1 \\ 1 \end{pmatrix} = \begin{pmatrix} 1/\sqrt{2} \\ 1/\sqrt{2} \end{pmatrix} \approx \begin{pmatrix} 0.707 \\ 0.707 \end{pmatrix}$$

**Paso 3: Hallar el Segundo Autovector $\mathbf{u}_2$ (asociado a $\lambda_2 = 4$)**
$$(C - \lambda_2 I) \mathbf{u}_2 = \mathbf{0} \quad \Rightarrow \quad \begin{pmatrix} 5-4 & 1 \\ 1 & 5-4 \end{pmatrix} \begin{pmatrix} u_{2,1} \\ u_{2,2} \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix} \quad \Rightarrow \quad \begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix} \begin{pmatrix} u_{2,1} \\ u_{2,2} \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$
La primera fila da: $u_{2,1} + u_{2,2} = 0 \quad \Rightarrow \quad u_{2,1} = -u_{2,2}$.
Un autovector no normalizado es $\begin{pmatrix} 1 \\ -1 \end{pmatrix}$. Normalizando:
$$\mathbf{u}_2 = \frac{1}{\sqrt{1^2+(-1)^2}} \begin{pmatrix} 1 \\ -1 \end{pmatrix} = \begin{pmatrix} 1/\sqrt{2} \\ -1/\sqrt{2} \end{pmatrix} \approx \begin{pmatrix} 0.707 \\ -0.707 \end{pmatrix}$$

**Resultado:**
$$\lambda_1 = 6, \quad \mathbf{u}_1 \approx \begin{pmatrix} 0.707 \\ 0.707 \end{pmatrix} \quad \text{(Primer CP)}$$
$$\lambda_2 = 4, \quad \mathbf{u}_2 \approx \begin{pmatrix} 0.707 \\ -0.707 \end{pmatrix} \quad \text{(Segundo CP)}$$

#### P3: Proyección y Reconstrucción 3D $\rightarrow$ 2D

**Datos:** Punto centrado $\mathbf{\tilde{x}} = \begin{pmatrix} 2 \\ 1 \end{pmatrix}$. Primer CP: $\mathbf{u}_1 = \begin{pmatrix} 1/\sqrt{2} \\ 1/\sqrt{2} \end{pmatrix}$.

**Paso 1: Proyectar el punto $\mathbf{\tilde{x}}$ sobre $\mathbf{u}_1$**
La coordenada proyectada $\mathbf{z}_1$ (la "carga" o *score*):
$$\mathbf{z}_1 = \mathbf{u}_1^T \mathbf{\tilde{x}} = \begin{pmatrix} 1/\sqrt{2} & 1/\sqrt{2} \end{pmatrix} \begin{pmatrix} 2 \\ 1 \end{pmatrix} = \frac{2}{\sqrt{2}} + \frac{1}{\sqrt{2}} = \frac{3}{\sqrt{2}}$$
$$\mathbf{z}_1 \approx 2.121$$

**Paso 2: Reconstruir el punto $\mathbf{\hat{x}}$ en 2D**
La reconstrucción $\mathbf{\hat{x}}$ se realiza usando la coordenada proyectada $\mathbf{z}_1$ y la dirección del componente principal $\mathbf{u}_1$:
$$\mathbf{\hat{x}} = \mathbf{z}_1 \mathbf{u}_1 = \frac{3}{\sqrt{2}} \begin{pmatrix} 1/\sqrt{2} \\ 1/\sqrt{2} \end{pmatrix} = \begin{pmatrix} 3/2 \\ 3/2 \end{pmatrix} = \begin{pmatrix} 1.5 \\ 1.5 \end{pmatrix}$$

**Paso 3: Calcular el Error de Reconstrucción (Distancia Euclidiana Cuadrada)**
$$\text{Error} = \| \mathbf{\tilde{x}} - \mathbf{\hat{x}} \|_2^2 = \left\| \begin{pmatrix} 2 \\ 1 \end{pmatrix} - \begin{pmatrix} 1.5 \\ 1.5 \end{pmatrix} \right\|_2^2$$
$$\text{Error} = \| \begin{pmatrix} 0.5 \\ -0.5 \end{pmatrix} \|_2^2 = (0.5)^2 + (-0.5)^2 = 0.25 + 0.25 = 0.5$$

**Resultado:**
La coordenada proyectada es $\mathbf{z}_1 \approx 2.121$.
El punto reconstruido es $\mathbf{\hat{x}} = \begin{pmatrix} 1.5 \\ 1.5 \end{pmatrix}$.
El error de reconstrucción es $0.5$.

#### P5: Varianza Explicada Acumulada

**Datos:** $\lambda_1 = 12.0, \quad \lambda_2 = 3.5, \quad \lambda_3 = 0.8, \quad \lambda_4 = 0.3$.

**Paso 1: Calcular la Varianza Total (Trace)**
$$\text{Varianza Total} = \sum_{i=1}^4 \lambda_i = 12.0 + 3.5 + 0.8 + 0.3 = 16.6$$

**Paso 2: Calcular la Varianza Explicada Acumulada para cada $k$**
Se utiliza la Varianza Explicada (Ecuación 3.1, implícita en la Diapositiva 10): $\frac{\lambda_{\text{acumulada}}}{\text{Varianza Total}}$.

    $$\frac{12.0}{16.6} \approx 0.7229 \quad \rightarrow \quad 72.29\%$$

    $$\frac{15.5 + 0.8}{16.6} = \frac{16.3}{16.6} \approx 0.9819 \quad \rightarrow \quad 98.19\%$$

**Paso 3: Determinar el Número Mínimo de Componentes para 90%**
El umbral deseado es $90\% = 0.90$.

  * Con $k=1$, la varianza explicada es $72.29\%$ (menor a $90\%$).
  * Con $k=2$, la varianza explicada es $93.37\%$ (mayor a $90\%$).

**Resultado:**
La varianza explicada acumulada es $72.29\%$ (k=1), $93.37\%$ (k=2), y $98.19\%$ (k=3). El número mínimo de componentes necesario para retener al menos el 90% de la varianza total es **$k=2$**.

### Soluciones Etapa 3 – Programación en Python para Google Colab
----
#### C1: Aplicación y Visualización (PCA Clásico)

**Objetivo:** Aplicar PCA en `load_wine`, visualizar 2D y determinar $k$ para 95% de varianza.

```python
# Cell 1: Importación de Librerías y Carga de Datos

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import seaborn as sns

# Cargar el dataset Wine
data = load_wine()
X = data.data
y = data.target
feature_names = data.feature_names
print(f"Dimensiones de los datos originales: {X.shape}")
```

```python
# Cell 2: Preprocesamiento (Estandarización)

# 1. Estandarización: PCA es sensible al escalado.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f"Datos escalados: Media ~{np.mean(X_scaled[0]):.2f}, Desv. Est. ~{np.std(X_scaled[0]):.2f}")
```

```python
# Cell 3: Gráfico de Varianza Explicada Acumulada (Selección de k)

# 2. Ajustar PCA (mantener todos los componentes para el análisis de varianza)
pca_full = PCA()
pca_full.fit(X_scaled)

# Calcular la Varianza Explicada Acumulada
explained_variance_ratio_cumsum = np.cumsum(pca_full.explained_variance_ratio_)
n_components = len(explained_variance_ratio_cumsum)
component_indices = np.arange(1, n_components + 1)

# Visualización del Gráfico de Codo
plt.figure(figsize=(8, 5))
plt.plot(component_indices, explained_variance_ratio_cumsum, marker='o', linestyle='-', color='b')
plt.axhline(y=0.95, color='r', linestyle='--', label='Umbral 95%')

# Encontrar el k para 95%
k_95_percent = np.argmax(explained_variance_ratio_cumsum >= 0.95) + 1 # +1 porque el índice es base 0

plt.axvline(x=k_95_percent, color='g', linestyle=':', label=f'k = {k_95_percent} (95%)')
plt.title('Varianza Explicada Acumulada (Scree Plot)')
plt.xlabel('Número de Componentes Principales')
plt.ylabel('Varianza Explicada Acumulada')
plt.xticks(component_indices)
plt.legend()
plt.grid(True)
plt.show()

print(f"Se necesitan k={k_95_percent} componentes para retener el 95% de la varianza.")
```

```python
# Cell 4: Visualización de las Dos Primeras Componentes

# 4. Reducir la dimensionalidad a k=2
pca_2d = PCA(n_components=2)
X_pca_2d = pca_2d.fit_transform(X_scaled)

# Creación de DataFrame para visualización
df_pca = pd.DataFrame(data=X_pca_2d, columns=['PC1', 'PC2'])
df_pca['Target'] = y

# Visualización 2D
plt.figure(figsize=(10, 8))
sns.scatterplot(
    x='PC1', y='PC2',
    hue='Target',
    palette=sns.color_palette("hsv", len(data.target_names)),
    data=df_pca,
    legend="full",
    alpha=0.8
)
plt.title('PCA 2D de Dataset Wine (Clases coloreadas)')
plt.xlabel(f'PC1 (Var. Explicada: {pca_2d.explained_variance_ratio_[0]*100:.2f}%)')
plt.ylabel(f'PC2 (Var. Explicada: {pca_2d.explained_variance_ratio_[1]*100:.2f}%)')
plt.grid(True)
plt.show()
```

**Salida Esperada (C1):** El código debe mostrar que $k=6$ o $k=7$ componentes son suficientes para el 95% de la varianza. La gráfica 2D mostrará una clara separación de las tres clases de vino, demostrando el poder de PCA para la visualización.

#### C2: Escalado y Optimización (Randomized/Incremental PCA)

**Objetivo:** Aplicar PCA optimizado para un dataset grande (MNIST), usando `svd_solver='randomized'` y `float32`.

```python
# Cell 1: Importación de Librerías y Carga de MNIST
# ¡Nota: Este proceso puede tardar unos minutos en descargar el dataset!
import time
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Cargar MNIST y forzar a float32 para ahorro de memoria
print("Descargando MNIST-784...")
X, y = fetch_openml('mnist_784', version=1, return_X_y=True, as_frame=False)
X = X.astype(np.float32)

print(f"Dimensiones de los datos originales (float32): {X.shape}")
```

```python
# Cell 2: Preprocesamiento y PCA Optimizado (Randomized SVD)

# 1. Estandarización
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Aplicar PCA con Randomized SVD (n_components=50)
k = 50
pca_randomized = PCA(n_components=k, svd_solver='randomized')

start_time = time.time()
X_pca_randomized = pca_randomized.fit_transform(X_scaled)
end_time = time.time()

# 3. Métricas
time_taken = end_time - start_time
explained_var_50 = np.sum(pca_randomized.explained_variance_ratio_)

print(f"\nPCA (Randomized SVD) finalizado en: {time_taken:.2f} segundos.")
print(f"Varianza Explicada Acumulada con k={k}: {explained_var_50*100:.2f}%")
print(f"Dimensiones de los datos reducidos: {X_pca_randomized.shape}")
```

```python
# Cell 3: Reconstrucción y Error (MSE)

# 4. Reconstrucción de la primera imagen (Índice 0)
sample_index = 0
original_image_scaled = X_scaled[sample_index, :]

# Reconstrucción: X_hat = Z_k * V_k^T + X_bar
# X_bar (media) es el atributo .mean_ del scaler.
# V_k^T (autovectores) es el atributo .components_ del PCA.
# Z_k (proyeccion) es X_pca_randomized

# Proyección de la muestra:
z_k = X_pca_randomized[sample_index, :]

# Reconstrucción del punto centrado:
reconstructed_scaled = pca_randomized.inverse_transform(z_k)

# De-escalado para obtener la imagen reconstruida (sumar la media original)
reconstructed_image = reconstructed_scaled * np.sqrt(scaler.var_) + scaler.mean_

# Cálculo del Error de Reconstrucción (MSE)
# Se calcula sobre la imagen desescalada para una métrica más interpretable
mse = np.mean((X[sample_index, :] - reconstructed_image) ** 2)

# Visualización de la imagen original vs. reconstruida
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(X[sample_index, :].reshape(28, 28), cmap='gray')
axes[0].set_title('Original')
axes[0].axis('off')

axes[1].imshow(reconstructed_image.reshape(28, 28), cmap='gray')
axes[1].set_title(f'Reconstruida (k={k})')
axes[1].axis('off')

plt.suptitle(f"Reconstrucción de MNIST. MSE: {mse:.2f}")
plt.show()

print(f"Error Cuadrático Medio de Reconstrucción (MSE) para la muestra {sample_index}: {mse:.2f}")
```
